# 1. Introduction:

This project investigates what player behaviours, specifically level of experience and age, impact whether or not a player will subscribe to the newsletter, created by Frank Wood and his research team in the computer science department at UBC. 

 The data set being analyzed is composed of 9 columns, two of which contain no values (individualId and organizationName), and 196 rows of observations. The columns within the data set include two numerical columns (Played_hours, age), three categorical columns (experience, name, gender), one binary variable (subscribe), and one identifying variable (hashedEmail). 

 Using this dataset, our research is focused on the question: Using the player's experience and age, can we predict if the player will subscribe to the newsletter? In answering this, our model will contribute to the greater research into players' characteristics and behaviours will dictate their subscription to the newsletter produced by Frank Wood. The columns of experience and age were selected because they are most reflective of the changes that can be made to the curation of a newsletter. The age of the players will inform writers of the level of writing and the detailed information necessary to fulfill the players' needs. On a similar note, the level of experience most common among readers will inform writers of the types of content they should include in their newsletter. For example, if most players are experienced, it wouldn’t make sense to include information on common knowledge in the game. Knowing this, the predictors' experience level and age are most suitable to determine subscription status of the new observations. 

```python
import pandas as pd
import altair as alt
import numpy as np
from sklearn import set_config
set_config(transform_output="pandas")
from sklearn.preprocessing import StandardScaler
from sklearn.compose import make_column_transformer
from sklearn.compose import make_column_selector
from sklearn.model_selection import GridSearchCV, cross_validate
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.metrics import recall_score, precision_score
```

### 1.1 Loading data and description

```python
players_url = "https://drive.google.com/uc?export=download&id=1Mw9vW0hjTJwRWx0bDXiSpYsO3gKogaPz"
players_messy=pd.read_csv(players_url)
players_messy
```

This table consists of 9 variables, where the last two are empty:

- The player's experience: whether they are just starting to play or are already a pro at the game, 
- whether they are subscribed to a newsletter related to the game or not, 
- their email address, which is transformed for privacy, 
- the number of hours they have played, 
- their name,
- their self-reported gender, 
- their self-reported age, 
- their individual ID and organization name, which are both empty for all observations. 

The dataset includes information from 196 players, containing both quantitative and qualitative data. Two columns, individual and organizationName, are empty and should be removed. Data was collected through in-game voice chats, and information was provided during sign-up. However, the collection method has limitations: much of the data is self-reported, making it subjective and potentially unreliable. Furthermore, requiring players to communicate vocally with strangers may deter those less comfortable with online interaction (e.g., introverted players), introducing bias toward more extroverted participants and reducing the dataset’s representativeness.

# 2 Methods

**To present a tidy, cohesive summary of our dataset, the team performed the following  wrangling steps**:
- load the data set into the shared notebook using a URL
- drop all columns excluding experience, subscribe, or age
- Reassign the categorical variable experience as numerical groups, to use KNN classification
- graph experience versus age, with colour-coded subscription status on either axis, to analyze a potential relationship
- filter the age range based on the largest grouping (to limit outliers)
- replot using these age ranges

**To begin the modelling of the data**:
- Create a preprocessor used to scale the variables
- fitting the preprocessor onto the wrangled dataset
- splitting the data into 75% training and 25% testing
- selecting a range of possible K values 
- fitting KNNClassifier and preprocessor into a pipeline (only using training set)
- performing a 5-fold grid search 
- plotting the accuracy and K-values (only using the training set)
- running said model on the test set, and finding the accuracy results
- creating a confusion matrix
- finally, plotting our prediction graph

### 2.1: Data Wrangling

```python
players_wrangled=players_messy.drop(columns=["hashedEmail", "name", "individualId", "organizationName", "gender", "played_hours"])
players_wrangled
```

```python
experience_map = {
    "Amateur": 1,
    "Beginner": 2,
    "Regular": 3,
    "Pro": 4,
    "Veteran": 5
}

players_wrangled["experience"] = players_wrangled["experience"].map(experience_map)
players_wrangled
```

### 2.1: Initial visdualizations

```python
experience_plot = alt.Chart(players_wrangled).mark_circle(size=60).encode(
    x=alt.X("experience:O", title="Experience level"),
    y=alt.Y("age:Q", title="Age"),
    color=alt.Color("subscribe:N", title="Subscribed"),
    tooltip=["experience", "age", "subscribe"]
).properties(
    title="Figure One: Age vs Experience by Subscription Status."
)

experience_plot
```

```python
age_plot=alt.Chart(players_wrangled).mark_circle().encode(
    x=alt.X("age").title("player_age"),
    y=alt.Y("experience").title("experience_level"), 
    color=alt.Color("subscribe").title("subscrition status")
).properties(
    title="Figure Two: experience vs age by Subscription Status."
)
age_plot
```

### 2.3: limiting outliers

```python
players_filtered = players_wrangled[(players_wrangled['age'] > 5) & (players_wrangled['age'] < 50)]
players_filtered
```

```python
filtered_experience_plot = alt.Chart(players_filtered).mark_circle(size=60).encode(
    x=alt.X("experience:O", title="Experience level"),
    y=alt.Y("age:Q", title="Age"),
    color=alt.Color("subscribe:N", title="Subscribed"),
    tooltip=["experience", "age", "subscribe"]
).properties(
    title="Figure Three: Age vs Experience by Subscription Status."
)

filtered_experience_plot
```

```python
filtered_age_plot=alt.Chart(players_filtered).mark_circle().encode(
    x=alt.X("age").title("player age"),
    y=alt.Y("experience").title("experience level"), 
    color=alt.Color("subscribe").title("subscrition status")
).properties(title='Figure Four: experience vs condensed age range by subscription status')
filtered_age_plot
```

### 2.4: Data analysis

```python
preprocessor = make_column_transformer(
    (StandardScaler(), make_column_selector(dtype_include="number")),
    remainder="passthrough",
    verbose_feature_names_out=False
)
preprocessor
```

```python
preprocessor.fit(players_filtered)
scaled_players = preprocessor.transform(players_filtered)
scaled_players
```

```python
scaled_age_plot=alt.Chart(scaled_players).mark_circle(opacity=0.5).encode(
    x=alt.X("age").title("player age"),
    y=alt.Y("experience").title("experience level"), 
    color=alt.Color("subscribe").title("subscribed or not")
).properties(title='Figure Five: experience vs age by subscription status (scaled)')
scaled_age_plot
```

```python
#split data 
players_train, players_test = train_test_split(players_wrangled, test_size=0.25, random_state=123, stratify=players_wrangled["subscribe"])

X_train = players_train[["experience", "age"]]
y_train = players_train["subscribe"]

X_test = players_test[["experience", "age"]]
y_test = players_test["subscribe"]
```

```python
#standardize training data 
players_preprocessor = make_column_transformer(
    (StandardScaler(), ["age","experience"]),
     verbose_feature_names_out=False
)
```

```python
#find best k using CV 
param_grid = {
    "kneighborsclassifier__n_neighbors": range(2, 15, 1),
}

players_pipe= make_pipeline(players_preprocessor, KNeighborsClassifier())

knn_tune= GridSearchCV(
    estimator=players_pipe, 
    param_grid=param_grid, 
    cv=5,
    n_jobs=-1,
    return_train_score=True)

knn_model= knn_tune.fit(X_train, y_train)

accuracies= pd.DataFrame(knn_model.cv_results_)
```

```python
#plot to visualize best k 
cross_val_plot = alt.Chart(accuracies).mark_line(point=True).encode(
    x=alt.X("param_kneighborsclassifier__n_neighbors")
        .title("Neighbors")
        .scale(zero=False),
    y=alt.Y("mean_test_score")
        .title("Mean Test Score")
        .scale(zero=False)
).properties(title='Figure Six: K values vs accuracy')
cross_val_plot
```

```python
#confirm best k value
knn_tune.best_params_
```

```python
#run the model on the test data using best k 
knn = KNeighborsClassifier(n_neighbors=11)

knn_pipeline = make_pipeline(players_preprocessor, knn)
knn_pipeline.fit(X_train,y_train)

players_test["predicted"] = knn_pipeline.predict(players_test[["experience", "age"]])

knn_pipeline.score(
    players_test[["experience", "age"]],
    players_test["subscribe"]
)
```

# 3. Results

```python
#create confusion matrix

precision_score(
    y_true=players_test["subscribe"],
    y_pred=players_test["predicted"],
    pos_label=True
)

recall_score(
    y_true=players_test["subscribe"],
    y_pred=players_test["predicted"],
    pos_label=True
)

pd.crosstab(
    players_test["subscribe"],
    players_test["predicted"]
)
```

**Precision= 0.77**

**Recall=1.00**

```python
#Scatter plot of experience versus age where background color indicates the decision of the classifier.

# 1. Create the grid of feature values
experience_grid = np.linspace(
    players_wrangled["experience"].min() * 0.95,
    players_wrangled["experience"].max() * 1.05,
    50
)

age_grid = np.linspace(
    players_wrangled["age"].min() * 0.95,
    players_wrangled["age"].max() * 1.05,
    50
)

grid = np.array(np.meshgrid(experience_grid, age_grid)).reshape(2, -1).T
grid = pd.DataFrame(grid, columns=["experience", "age"])


# 2. Predict at the grid points
knnPredGrid = knn_pipeline.predict(grid) #check this

# 3. Bind predictions to the grid
prediction_table = grid.copy()
prediction_table["predicted"] = knnPredGrid


# 4. Plot original data
unscaled_plot = alt.Chart(players_wrangled).mark_point(
    opacity=0.6,
    filled=True,
    size=40
).encode(
    x=alt.X(
        "experience",
        scale=alt.Scale(
            nice=False,
            domain=[
                players_wrangled["experience"].min() * 0.95,
                players_wrangled["experience"].max() * 1.05
            ]
        )
    ),
    y=alt.Y(
        "age",
        scale=alt.Scale(
            nice=False,
            domain=[
                players_wrangled["age"].min() * 0.95,
                players_wrangled["age"].max() * 1.05
            ]
        )
    ),
    color=alt.Color("subscribe:N", title="Subscribe")
)


# 5. Plot prediction background
prediction_plot = alt.Chart(prediction_table).mark_point(
    opacity=0.05,
    filled=True,
    size=300
).encode(
    x="experience",
    y="age",
    color=alt.Color("predicted:N", title="Subscribe")
)


# 6. Combine plots
final_plot=unscaled_plot + prediction_plot.properties(title='Figure Seven: Prediction plot for subscription status using age and experience')
final_plot
```

# 4. Discussion and conclusion


The goal of this project was to determine if a player’s age and level of experience can predict subscription to the Minecraft newsletter. Using a k-Nearest Neighbours (k-NN) classifier, we trained and tuned a model based on these two predictor variables. The final model, selected via cross-validation, yielded k = 11 and an overall test-set accuracy of approximately 77.6%. This indicates that age and experience contain meaningful information about subscription behaviour. We also found the precision value to be 77%, and a recall value of 100%. Since the recall is such a high value, the data set likely struggles with a class imbalance, leading to there always being a prediction of subscribed. This high accuracy value may also be accounted for by class imbalance. 

**Expected Findings vs. Outcome:**

Before conducting the analysis, we hypothesized that:

- Younger players would be more likely to subscribe, since they may be more curious about the game or more motivated to learn.
- Less experienced players might also subscribe at higher rates, as newsletters can offer tips, updates, and guidance.

The results closely matched these expectations. The decision-boundary visualizations showed that subscription was much more common among players in their teens and early twenties, especially those with moderate or low experience levels. Meanwhile, highly experienced players—particularly “Veteran” players—were less likely to subscribe, likely because they have less need for beginner-level resources.

The confusion matrix supports this interpretation. The classifier correctly predicted most subscribed players (36 true positives) and a smaller number of non-subscribers (2 true negatives). Its main source of error was misclassifying unsubscribed players as subscribed, reflecting the dataset’s imbalance—most players in the sample are subscribers, and the model learned this bias.

**Potential Impact of These Findings**

These findings have practical implications for how the research team might shape newsletter content or recruitment strategies:
- Newsletter content can be tailored toward younger or less experienced players, who appear more enthusiastic about subscribing.
- More advanced content could be developed to better engage highly experienced players, who are currently less likely to subscribe.
Recruitment strategies may emphasize inviting new or younger players, as they demonstrate higher interest.
Understanding which demographic groups engage with the newsletter helps the research team create more relevant, inclusive, and effective communication.

**Limitations**

Several limitations should be considered when interpreting our results:

- Small dataset (n = 196): With limited observations, model estimates may be sensitive to noise.
Self-reported data: Variables such as age and experience may be misreported or inconsistently interpreted.
- Class imbalance: The majority of players subscribed to the newsletter, influencing the model toward predicting “True.”
- k-NN limitations: The method relies on scaled numerical distances, which may oversimplify complex relationships between age, experience, and subscription behaviour. KNN performance may also struggle with class imbalance, depending on the variables assigned.
Despite these constraints, the k-NN classifier performed reasonably well and aligned with our expectations.


**Future Questions**
Our analysis suggests several directions for further investigation:
- Would including additional predictors—such as play time, gender, or activity frequency—improve predictive accuracy?
- How would other models (e.g., logistic regression, decision trees, random forests) compare to k-NN in performance and interpretability?
- Would a larger and more balanced dataset yield stronger or different patterns of association?

Exploring these questions could deepen understanding of player engagement and support more effective communication strategies for the PLAICraft research team.
